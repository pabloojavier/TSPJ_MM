```python
import tsplib95
import math
import numpy as np

instancia = '/Users/pgutiea/Desktop/TSPJ_MM/Codigos/Data/tsp/eil51.tsp'
problem = tsplib95.load(instancia)
n = len(list(problem.get_nodes()))

points = [(problem.node_coords[i][0],problem.node_coords[i][1]) for i in range(1,n+1)]
nodos = [i for i in range(n)]
arcos = [(i,j) for i in nodos for j in nodos if i !=j]
dist = {(i, j):
        (math.sqrt(sum((points[i][k]-points[j][k])**2 for k in range(2))))
        for i in range(n) for j in range(i)}
eil51 = np.zeros((n,n))
for i,j in dist.keys():
    eil51[i,j] = dist[i,j]
    eil51[j,i] = dist[i,j]
```

```python
c = eil51
n = len(c)
r = 0         # start tour at an arbitrary root node 'r'

import networkx as nx

# Complete graph on n vertices
G = nx.complete_graph(n)

# DG is directed version of G. Replaces each edge {i,j} by (i,j) and (j,i)
DG = nx.DiGraph(G)
```

```python
import gurobipy as gp
from gurobipy import GRB
```

```python
# Create model object
m = gp.Model()

# Create variable for each edge of DG
x = m.addVars( DG.edges, vtype=GRB.BINARY )

# Objective function: minimize cost of tour
m.setObjective( gp.quicksum( c[i][j] * x[i,j] for i,j in DG.edges ), GRB.MINIMIZE )

# Leave each city once
m.addConstrs( gp.quicksum( x[i,j] for j in DG.neighbors(i) ) == 1 for i in DG.nodes )

# Enter each city once
m.addConstrs( gp.quicksum( x[i,j] for i in DG.neighbors(j) ) == 1 for j in DG.nodes )

m.update()
```

```python
from cortes_github import all_functions as cuts
for i in cuts:
    print(i.__name__)
```

```python
# Solve the MIP using an integer separation callback
corte = cuts[0]
print(f'Optimizing with cut {corte.__name__}...\n')
m.Params.LazyConstraints = 1
m._callback = corte
m._x = x
m._epsilon = 0.000000000001
m._DG = DG
m.optimize(m._callback)
```

```python
print("Objective:",m.objVal)

epsilon = 0.000000000001
xval_sparse = { (i,j) : m._x[i,j].x for i,j in DG.edges if m._x[i,j].x > epsilon }
print("xval =",xval_sparse)
```
